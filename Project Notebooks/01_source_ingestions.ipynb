### Load Library

```python
import re
```

###Define path and file names

```python
input_path = '/Volumes/fifa_bi_dev/default/source_files'
files = ['world_cup_matches.csv',
             'worldcups.csv',
             'average-monthly-surface-temperature.csv']




output_path = 'fifa_bi_dev.bronze_schema'
```

### GET TABLE FROM FILENAME

```python
table_dic = {
    file: file.replace(" ", "_").replace("-", "_").removesuffix(".csv")
    for file in files
}
```

## Load Functions

```python
def load_csv_files(i_path, table_dic, o_path):

    for key, value in table_dic.items():
        df = spark.read.format("csv").option("header", True).load(f'{i_path}/{key}')
        print(f'file_name: {key} is loaded')
        df = df.toDF(*[
                    re.sub(r'[^a-zA-Z0-9_]', '_', c).lower()
                    for c in df.columns
                    ])

        # load the files in a table
        df.write.format("delta").mode("overwrite").saveAsTable(f'{o_path}.{value}')
        print(f'table_name: {value} is updated') 
        print('-------------------') 

```

## RUN FUNCTIONS

```python
load_csv_files(input_path,table_dic, output_path)
```
