```python
schema = 'fifa_bi_dev.silver_schema'
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window

df = spark.read.table('fifa_bi_dev.bronze_schema.average_monthly_surface_temperature')

result = df.withColumn('country', col('entity') ).drop('entity')
result = result.withColumn('country_code', col('code') ).drop('code')
result = result.withColumn('monthly_average', col('monthly_average').cast('double'))
result = result.withColumn('avg_temperature', round(col('monthly_average'),1)).drop('code')

display(result.limit(20))
```

```python
result.write.format("delta").mode("overwrite").saveAsTable(f'{schema}.silver_monthly_temperature')
print(f'table_name: monthly_temperature is updated')
```
