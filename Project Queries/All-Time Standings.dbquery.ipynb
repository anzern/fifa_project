```sql
select team_code, team_country, count(distinct tournament_name) worldcup_appearance, 
count(case when ms.rank = 1 and result = 'Win' then match_id end) as 1st,
count(case when ms.rank = 1 and result = 'Lost' then match_id end) as 2nd,
count(case when ms.rank = 2 and result = 'Win' then match_id end) as 3rd,
count(case when ms.rank = 2 and result = 'Lost' then match_id end) as 4th,
count(case when result = 'Win' then result end ) W,
count(case when result = 'Draw' then result end ) D,
count(case when result = 'Lost' then result end ) L,
sum(goals_for) GF,
sum(goals_against) GA,
sum(goals_for) - sum(goals_against) GD

from fifa_bi_dev.gold_schema.gold_team_all_time_standing s
left join fifa_bi_dev.gold_schema.dim_match_stage ms on ms.round = s.round and ms.match_stage = s.match_stage
group by 1,2
order by 3 desc, 2
```
