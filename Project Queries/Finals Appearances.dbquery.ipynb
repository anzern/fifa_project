```sql
select team_country, 
count(distinct match_id ) Finals,
count(distinct case when result = 'Win' then match_id end) W,
count(distinct case when result = 'Lost' then match_id end) L,
concat_ws(', ', sort_array(collect_list(CAST(year AS STRING)),false)) AS Years
from fifa_bi_dev.gold_schema.gold_team_all_time_standing s
where round = 'Final'
group by 1
order by 2 desc
```
