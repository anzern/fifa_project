```sql
-- TOP 20 Biggest Wins

with t as (
select Year, 
team_country Team_1,
opponent_country Team_2,
concat(goals_for,'-',goals_against) Score, 
Round

from fifa_bi_dev.gold_schema.gold_team_all_time_standing
where result in ('Win', 'Draw')
order by goals_for + goals_against desc, goals_for desc
)

select 
Year,
Team_1,
Team_2,
Score,
Round
 from t
 limit 20
```
