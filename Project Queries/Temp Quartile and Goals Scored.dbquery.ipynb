```sql

WITH quartiles AS (
    SELECT
        *,
        NTILE(4) OVER (ORDER BY adj_temperature_final) AS quartile
    FROM fifa_bi_dev.gold_schema.gold_match_temperature
)

SELECT
    quartile,
    COUNT(*) AS total_matches,
    concat(MIN(adj_temperature_final), ' - ', MAX(adj_temperature_final)) AS temp_range,
    round(SUM(team_1_goals + team_2_goals)/COUNT(*),1) avg_goals 
FROM quartiles
GROUP BY quartile
ORDER BY quartile;
```
